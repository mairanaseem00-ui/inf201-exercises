```python
# <Your code here>

# Format for report:
# (Note: These are example numbers, not the expected values for the input given below)
#
# --- SENSOR REPORT: SENSOR NAME ---
# Valid readings      :    89.3%
# Average reading     :     5.01
# Removed readings    :        3

# Input for demonstration
sensor_name = "sensor a"
sensor_readings = [2.673, -3.876, 4.589, None, None, None, -6.111, 0.435, None, 0.005,
                   2.444, 10.897, 0.019, 10.222, -0.543, None, 8.097, 0.962, None, 1.005,
                   None, 0.765, 0.765, 0.765, 0.765, 7.937, -1.007, 4.405, None, None]
sensor_threshold = 0.010
```

```python




# Cleaning data by removing None and values less then threshold

def clean_sensor_reading (sensor_readings: list [float | None],sensor_threshold: float) -> list[float]:
    """
    Remove invalid sensor readings.

    A reading is valid if it is not None and is greater
    than or equal to the threshold.

    Args:
        sensor_readings: List of sensor readings.
        sensor_threshold: The minimum value for a valid reading.

    Returns:
        A list with values that are valid readings.
    """
# Making a list to add values that are valid readings.
    cleaned = []

    for reading in sensor_readings:
        if reading is not None and reading >= sensor_threshold:
            cleaned.append(reading)
    return cleaned
    

# Checking results to see if list is cleaned 
result_clean = clean_sensor_reading(sensor_readings, sensor_threshold)
display(result_clean)
print(len(result_clean))

```

```python
def sensor_report (sensor_name: str, sensor_readings: [float|None], result_clean: list[float]) -> str:
    """
    Creating a report to read of information about sensor readings 

    Args:
        sensor_name: The name of the sensor.
        sensor_readings: The real/orignal list of sensors.
        result_clean: The list of sensors after cleaned and gives valid sensor readings.

    Returns:
        A string containing that sensor report.
    """
    # calculating valid report results 
    valid_reading_count = len(result_clean)
    avg_reading = sum(result_clean)/ valid_reading_count
    valid_reading_percent = valid_reading_count / len(sensor_readings)*100
    removed_reading = len(sensor_readings) - valid_reading_count

    # Making a visual report with data found 
    report = f"""--- SENSOR REPORT: {sensor_name} ---
Valid readings      : {valid_reading_percent:.1f}%
Average reading     : {avg_reading:.2f}
Removed readings    : {removed_reading}
"""

print(report)
```
