## Class Overview

This project implements a mini role-playing game (RPG) battle simulator using object-oriented programming in Python. The game includes six classes: Character, Dice, Weapon, Warrior, Mage, and Battle.

### Character

The Character class represents a basic fighter.

- **Attributes:** `name`, `_hp`, `hp_max`, `damage`, `weapon`
- **Methods:** `__init__()`, `take_damage()`, `heal()`, `attack()`, `introduce()`, `__repr__()`, `__str__()`
- **Properties:** `hp`, `is_alive`

The `_hp` attribute stores the character's current health. The read-only properties allow access to health and alive status without directly modifying `_hp`.

### Dice

The Dice class represents a die with a configurable number of sides.

- **Attribute:** `sides`
- **Methods:** `__init__()`, `roll()`, `__repr__()`, `__call__()`

This class generates random numbers used to calculate weapon damage.

### Weapon

The Weapon class represents an item that a character can equip.

- **Attributes:** `name`, `damage_dice`
- **Methods:** `__init__()`, `roll_damage()`

Each weapon contains a Dice object, demonstrating object composition.

### Warrior

The Warrior class inherits from Character and specializes in defense.

- **Additional attribute:** `armor`
- **Overridden method:** `take_damage()`

The Warrior reduces incoming damage by its armor value before applying damage to its HP.

### Mage

The Mage class inherits from Character and specializes in magical attacks.

- **Additional attributes:** `mana`, `mana_max`
- **Overridden method:** `attack()`

When sufficient mana is available, the Mage deals double base damage using magic. Otherwise, it performs a normal attack and restores some mana.

### Battle

The Battle class manages combat between two characters.

- **Attributes:** `hero`, `enemy`
- **Methods:** `__init__()`, `play_round()`, `simulate_battle()`

The Battle class manages turn-based combat until one character is defeated.

### Object-Oriented Programming Concepts

The project demonstrates encapsulation through `_hp` and read-only properties, inheritance through Warrior and Mage, polymorphism through overridden methods, and composition through Character–Weapon and Weapon–Dice relationships.

## Task 5: Running battles

Create a `Battle` class to manage turn-based combat between two `Character` objects.

* Initialize `Battle` with `hero: Character` and `enemy: Character`.
* Implement a `play_round()` method where `hero` attacks `enemy`, and if `enemy` is still alive, `enemy` attacks `hero`.
* Implement a `simulate_battle()` method with a `while` loop that runs rounds until one character is no longer alive.

```python
# Creating two characters
hero = Character(name="Henrik", hp_max=30, damage=10)
enemy = Character(name="Ivar", hp_max=20, damage=10)

# Creating a battle
battle = Battle(hero=hero, enemy=enemy)

# Testing one round
battle.play_round()

print(f"{hero.name} has {hero.hp} HP left")
print(f"{enemy.name} has {enemy.hp} HP left")

# Testing the full battle
battle.simulate_battle()

print(f"{hero.name} has {hero.hp} HP left")
print(f"{enemy.name} has {enemy.hp} HP left")
```

### Task 5

The results show that the Battle class works as expected. In the first round, Henrik attacks Ivar, reducing his HP from 20 to 10. Since Ivar is still alive, he attacks Henrik, reducing his HP from 30 to 20. The simulate_battle() method continues the battle using a while loop. In the next round, Henrik attacks again, reducing Ivar's HP to 0. Since Ivar is no longer alive, he cannot attack back, and the battle ends. Henrik wins with 20 HP remaining. This confirms that both play_round() and simulate_battle() work correctly.

```python
# Simulating a battle between a Warrior and a Mage

warrior = Warrior(
    name="Henrik",
    hp_max=100,
    damage=12,
    armor=5
)

mage = Mage(
    name="Ivar",
    hp_max=80,
    damage=10,
    mana=30
)

battle = Battle(hero=warrior, enemy=mage)
battle.simulate_battle()

print(f"{warrior.name} has {warrior.hp} HP remaining")
print(f"{mage.name} has {mage.hp} HP remaining")

if warrior.is_alive:
    print(f"{warrior.name} wins!")
else:
    print(f"{mage.name} wins!")
```

### Warrior vs Mage Battle – Discussion

In this simulation, we tested a battle between a Warrior and a Mage without equipped weapons.

The Warrior has 5 armor points, which reduce incoming damage. The Mage initially uses magical attacks that deal double its base damage and consume 10 mana per spell.

When the Mage has insufficient mana, it switches to its normal attack and restores 5 mana.

The battle continues until one character reaches 0 HP. This example demonstrates how inheritance and method overriding allow different character types to behave differently during combat.

```python
# Battle between a Warrior and a Mage with weapons

sword = Weapon("Sword", Dice(12))
staff = Weapon("Magic Staff", Dice(8))

warrior = Warrior(
    name="Henrik",
    hp_max=100,
    damage=12,
    weapon=sword,
    armor=5
)

mage = Mage(
    name="Ivar",
    hp_max=80,
    damage=10,
    weapon=staff,
    mana=30,
    mana_max=30
)

# Create and simulate the battle
weapon_battle = Battle(hero=warrior, enemy=mage)
weapon_battle.simulate_battle()

# Display final results
print(f"{warrior.name}: {warrior.hp} HP remaining")
print(f"{mage.name}: {mage.hp} HP remaining")
print(f"{mage.name}: {mage.mana} mana remaining")

if warrior.is_alive:
    print(f"Winner: {warrior.name}")
else:
    print(f"Winner: {mage.name}")
```
